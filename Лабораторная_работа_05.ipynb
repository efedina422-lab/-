## Задание

В файле pr5_3.txt содержится последовательность целых чисел.
Элементы последовательности могут принимать целые значения от
–10 000 до 10 000 включительно. Определите и запишите в ответе
сначала количество пар элементов последовательности, в которых
разность элементов чётна и кратна 37, затем максимальную из сумм
элементов таких пар. В данной задаче под парой подразумевается
два идущих подряд элемента последовательности.

## Решение

Файл pr5.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

```python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract5/pr5_3.txt
```

Можно скопировать и поместить файл с данными любым удобным способом

Реализуем чтение строк из файла и конвертацию отдельных строк в числовые значения. Каждое получаемое число будем добавлять в список `data`.

```python
f = open('pr5_3.txt')
a = [int(x) for x in f]
k = 0
mx = -20001
for i in range(len(a) - 1):
    if (a[i+1] - a[i]) % 2 == 0 and (a[i+1] - a[i]) % 37 == 0:
        k += 1
        mx = max(mx, a[i] + a[i+1])
print(k, mx)
```
