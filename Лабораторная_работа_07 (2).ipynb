## Обработка символьной последовательности

Строки, подобно списка, являются коллекциями. Только в качестве элементов последовательности выступают символы. Для строк также используются функции нахождения длины `len()`, операция индексации `[]`

Например, можно посимвольно обходить строку, используя цикл `for` и индекс строки:

```python
import csv

!wget -blob:https://github.com/e4ff376b-70d0-4a53-b5d3-bf1f39d66595
products = []
with open('p7_data_26.csv', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        # Убираем знак '$' в начале цены и переводим в дробное число
        row['Price'] = float(row['Price'][1:])
        # Переводим количество и скидку в целые числа
        row['Count'] = int(row['Count'])
        row['discount'] = int(row['discount'])

        products.append(row)

print("Таблица p7_data_26.csv успешно загружена. Всего товаров:", len(products))

```

## Решение задачи 1

Будем посимвольно обходимть строку и накапливать символы-цифры в строке `target`. Если цифровая последовательность прервется, то конвертируем `target` в число `int` и проверяем является ли число четным и не превысило ли оно максимум

```python

categories = {'A': 0, 'B': 0, 'C': 0, 'D': 0}

for item in products:
    cat = item['Category']
    categories[cat] += 1

print("Количество товаров по категориям:")
for cat, count in categories.items():
    print(f"Категория {cat}: {count} шт.")

```

## Решение задачи 2

Решение данной задачи осуществляется аналогично предыдущей. Можно даже вывести самуюдлинную буквенную строку

```python
suppliers = {}

for item in products:
    company = item['Company']
    if company == '':
        company = 'Не указан'
    volume = item['Price'] * item['Count']
    if company not in suppliers:
        suppliers[company] = 0
        suppliers[company] += volume

print("Суммарный объём товаров по поставщикам:")
for company, total in suppliers.items():
    print(f"{company}: ${total:.2f}")
```

## Решение задачи 3

Для решения данной задачи будем использовать коллекцию словарь `dict()`. Данная коллекция содержит элемкнты в виде пар `ключ : значение`. Доступ к элементам осуществляется по ключу с использованием операции ндексации.

Заполним словарь d, который после прохода по строке должен принять вид:
d = {'A' : 10, 'B' : 127, '3' : 217, ... }
то есть в качестве ключа будем использовать символ, а вкачестве значения - количество таких символов в строке:

```python
max_price = -1
best_item = None

for item in products:
    # Вычисляем цену после применения скидки
    final_price = item['Price'] * (100 - item['discount']) / 100

    if final_price > max_price:
        max_price = final_price
        best_item = item

print("Самый дорогой товар с учётом скидки:")
print(f"Товар: {best_item['Name']} (ID: {best_item['ID']})")
print(f"Цена без скидки: ${best_item['Price']:.2f}, скидка: {best_item['discount']}%")
print(f"Итоговая цена со скидкой: ${max_price:.2f}")
```

В получившемся словаре необходимо найти самый частотный символ. Для этого не будем обходить словарь, а просто воспользуемся встроенной функцией `max()`, которую применим к коллекции `values()` словаря

##Задание 4.

```python
min_count = products[0]['Count']
min_item = products[0]

for item in products:
    if item['Count'] < min_count:
        min_count = item['Count']
        min_item = item

print("Товар с минимальным количеством:")
print(f"Товар: {min_item['Name']} (ID: {min_item['ID']})")
print(f"Количество на складе: {min_count} шт.")
```

14.  Разница между методами pop() и popitem()
pop(key) — удаляет элемент по указанному ключу и возвращает его значение.
popitem() — не принимает аргументов, удаляет последнюю добавленную пару «ключ : значение» и возвращает ее в виде кортежа (ключ, значение).
15.Что возвращает метод items()?
Он возвращает специальный объект представления (dict_items), содержащий все пары словаря в виде списка кортежей: [(ключ1, значение1), (ключ2, значение2), ...]. Удобен для одновременного перебора ключей и значений в цикле for.
16. Что возвращают методы keys() и values()?
keys() — возвращает представление (dict_keys) всех ключей словаря.
values() — возвращает представление (dict_values) всех значений словаря.
