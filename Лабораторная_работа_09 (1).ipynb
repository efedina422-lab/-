```python
import math
class Pyramid:
  def __init__(self, a=1.0, h=1.0):
        self.a = float(a)
        self.h = float(h)

  def InputData(self):
        self.a = float(
            input("Введите сторону основания a: ").replace(",", ".")
        )
        self.h = float(input("Введите высоту пирамиды h: ").replace(",", "."))

  def PrintData(self):
        print(f"Сторона основания a = {self.a}")
        print(f"Высота пирамиды h = {self.h}")

  def GetS(self):
        s_osn = (self.a**2 * math.sqrt(3)) / 4
        apofema = math.sqrt(self.h**2 + (self.a**2) / 12)
        s_bok = (3 * self.a * apofema) / 2
        return s_osn + s_bok

  def GetV(self):
        s_osn = (self.a**2 * math.sqrt(3)) / 4
        return (1 / 3) * s_osn * self.h

  def Info(self):
    print("*" * 30)
    print("Правильная треугольная пирамида")
    self.PrintData()
    print(f"Площадь поверхности: {self.GetS():.3f}")
    print(f"Объем: {self.GetV():.3f}")
    print("*" * 30)



p1 = Pyramid(6, 10)
p1.Info()

p2 = Pyramid()
p2.InputData()
p2.Info()


```

1. Класс — это пользовательский тип данных, на основе которого создаются конкретные объекты.
3. Назначение: Статические методы используются для выполнения вспомогательных задач, связанных с классом по смыслу, но не требующих доступа к данным конкретного объекта.
Отличия: 1. Обычно метод принимает первым параметром self, а статический метод self не принимает. 2. Статический метод можно вызвать напрямую через имя класса, не создавая объект (например, Pyramid.Load(...)). 3.Статический метод не может напрямую изменять поля конкретного экземпляра.
4. Параметр self — это ссылка на конкретный текущий объект класса, для которого в данный момент вызывается метод. С помощью self метод определяет, к полям какого именно объекта нужно обратиться.
5. В Python внутри одного класса технически может быть только один действующий метод __init__. Если объявить в классе несколько методов с именем __init__, каждый следующий просто перезапишет предыдущий, и работать будет только самый последний. Для создания объектов с разным набором параметров используются значения по умолчанию.
6. Чтобы сделать атрибут приватным, перед его именем ставят два нижних подчеркивания (__), например: self.__secret_value = 10.
7. Задача специального метода __str__() — вернуть читаемое строковое представление объекта, понятное пользователю.
8. Метод __str__() вызывается автоматически в следующих случаях:
• При передаче объекта в стандартную функцию вывода print(obj).
• При явном преобразовании объекта в строку с помощью функции str(obj).
9. Атрибут экземпляра — это переменная, которая принадлежит конкретному созданному объекту и хранит его индивидуальные данные. Обычно создается внутри __init__ через self (например, self.a = 5). У каждого объекта значения этих атрибутов свои. 10. Эти методы используются для безопасной работы со скрытыми полями класса: • Метод-получатель (геттер / getter) — метод, который только возвращает (считывает) значение скрытого поля, не изменяя его.
• Метод-мутатор (сеттер / setter) — метод, который изменяет

```python
#2
#Метод __init__ является конструктором класса.
#Он вызывается автоматически в момент создания нового объекта и используется для начальной инициализации его полей данных.
class Box:

    def __init__(self, width, height):
        self.width = width
        self.height = height

b = Box(10, 20)
print(b.width, b.height)
```
