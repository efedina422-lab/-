## Решение

Файл `pr7_00.txt` расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

Можно скопировать и поместить файл с данными нв виртуальную машину Google Colab любым удобным способом

Реализуем чтение строк и их преобразование в словари:

```python
# список для хранения пользователей
peoples = []
with open('p7_data_00.csv') as f:
  f.readline()
  for line in f:
    d = line.strip().split(',')
    # формируем словарь для пользователя
    user = {}
    # ID,Fname,Lname,Email,Gender,Salary,HasChildren
    user['ID'] = d[0]
    user['Fname'] = d[1]
    user['Lname'] = d[2]
    user['Email'] = d[3]
    user['Gender'] = d[4]
    user['Salary'] = float(d[5][1:])
    user['HasChildren'] = True if d[6]=='true' else False
    peoples.append(user)
print(len(peoples))
```

Посмотрим на формат получившегося элемента списка:

```python
print(peoples[0])
```

Решим задачу: демонстрация возможностей сортировки пользователей. Будем сортировать по уровню заработной платы

```python

sorted_users = sorted(peoples, key=lambda u: u['Salary'], reverse=True)

print("ТОП-5 самых высоких зарплат:")
for u in sorted_users[:5]:
    print(f"{u['Fname']} {u['Lname']} — ${u['Salary']}")
```

Решим задачу: фильтрация пользователей. Найдем всех пользователей с зарплатой выше 5000.00, мужского пола и без детей

```python
filtered_users = list(filter(
    lambda u: u['Gender'] == 'Female' and u['HasChildren'] and u['Salary'] > 8000,
    peoples
))

print(f"Найдено подходящих пользователей: {len(filtered_users)}")
for u in filtered_users[:5]:
    print(f"{u['Fname']} {u['Lname']} | Дети: {u['HasChildren']} | Зарплата: ${u['Salary']}")
```

Решим задачу: продемонстрировать возможности отображения. На основе имеющегося списка построить новый, который содержит только поля ФИО и зарплата

```python

mapped_users = list(map(
    lambda u: f"{u['Fname']} {u['Lname']} ({u['Email']}) -> Годовой доход: ${u['Salary'] * 12:.2f}",
    peoples
))

print("Первые 5 карточек с годовым доходом:")
for card in mapped_users[:5]:
    print(card)
```

1. Анонимная функция (или лямбда-функция) — это небольшая однострочная функция в Python, которая объявляется без имени с помощью ключевого слова lambda. Она не сохраняется в памяти под постоянным именем и используется непосредственно в том месте, где была создана.
2. Синтаксис выглядит следующим образом: lambda аргументы: выражение. lambda — ключевое слово, обозначающее начало анонимной функции.аргументы — входные параметры, перечисленные через запятую. выражение — единственное действие, результат которого автоматически возвращается функцией.
3. Назначение: Лямбда-выражения нужны для сокращения кода в ситуациях, когда требуется передать простую функцию в качестве аргумента в другую функцию (например, в map(), filter(), sort() или sorted()), а создавать для этого полноценную именованную функцию через def нецелесообразно.
