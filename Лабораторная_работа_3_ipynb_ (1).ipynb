## Задание

Напишите функцию, которая возвращает True, если переданное ей
число является квадратом некоторого натурального числа, в
противном случае – False

## Решение

### Объявление функции

```python
def chislo(n):
    if n < 1:
        return False
    k = 1
    while k * k < n:
        k += 1
    return k * k == n
print(chislo(16))
```

### Демонстрация использования функции

```python
print(chislo(16))
print(chislo(14))
print(chislo(25))
print(chislo(31))
```
