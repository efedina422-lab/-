## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $w, h, b, q3, y$. Переменная, которая должна быть вычислена на основе независимых: $g$.

```python
w=10
h=30.67
b=5
q3=5
y=2
print("w = ", w, ", type(w) = ", type(w))
print("h = ", h, ", type(h) = ", type(h))
print("b = ", b, ", type(b) = ", type(b))
print("q3 = ", q3, ", type(q3) = ", type(q3))
print("y = ", y, "type(y) = ", type(y))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $Z$:

```python
g=(w**h-w**(1/3))/(q3*(b+120*abs(w-h)))
print('g=',g)
print('Type of g: ', type(g))
```
