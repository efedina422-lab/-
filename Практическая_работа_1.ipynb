```python
a = True
b = False
c = True
result = a + b + c
print(result)
print(type(result))
```

bool в Python наследуется от int: True = 1, False = 0. При сложении происходит неявное приведение к int, поэтому 1 + 0 + 1 = 2, а тип результата — <class 'int'>

##Контрольные вопросы

1. Переменная в Python- имя-ссылка на объект в памяти. При присваивании имя связывается с объектом. Тип хранится в объекте, а не в имени. Корректные имена: count, i2, _x2. Некорректное: 2value- нельзя начинать с цифры.

2. Основные типы: int (42), float (3.14), str ("Привет"), bool (True).Типизация динамическая, потому что тип определяется во время выполнения и может меняться у одной и той же переменной.

3. print() выводит объекты в стандартный поток, преобразуя их в строки, разделяя пробелом и завершая переводом строки. Способы: одна переменная; несколько через запятую (авто-пробелы); f-строка (полный контроль формата). Разница в выводе: при перечислении через запятую Python сам вставляет пробелы; f-строка даёт полный контроль над форматом

4. type() возвращает класс объекта. Например, type(10) <= <class 'int'>, type("hi") <= <class 'str'>, type(False) <= <class 'bool'>. Запись читается как «объект принадлежит классу …».

6. / — истинное деление, всегда float (7/2 = 3.5). // — целочисленное с округлением вниз (-7//2 = -4, а не -3).

7. Ошибка (ValueError) возникает при преобразовании нечисловой строки, например int("abc") или int("3.14").

8. Любое выражение возвращает объект, который можно присвоить или передать в print().

9. Сравнения: ==, !=, <, >, <=, >=. Логические: and, or, not. Приоритет: арифметика → сравнения → not → and → or.
10. Разные типы: Python неявно повышает тип (int + float → float, bool + int → int). Python не преобразует str в int автоматически — это осознанное решение, чтобы избежать неоднозначностей.

11. Способы вывода: через запятую (быстро, авто-пробелы); f-строка (удобна для форматирования и вычислений); .format() (шаблоны, старый код); конкатенация (только строки, требует str()).
